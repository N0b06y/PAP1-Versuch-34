```python
import numpy as np
import uncertainties
from uncertainties import unumpy as un
from uncertainties import ufloat
```

```python
# Takes an array with uncertainties assinged to each value
# and puts out the meanValue with the 'properly propegated error'
# (Taking the means deviation and it's propegated error, which are added quadratically)
def errorMean(foo):
    k = len(foo)
    unMean = foo.mean()
    meanDev = muStd(foo)
    stdDev = ufloat(0, meanDev)
    return (unMean + stdDev)


# Compares two values on their sigma domain
def sigmaDev(foo, bar):
    foo = foo if isinstance(foo, uncertainties.UFloat) else ufloat(foo, 0)
    bar = bar if isinstance(bar, uncertainties.UFloat) else ufloat(bar, 0)
    return np.abs(foo.nominal_value - bar.nominal_value)/((foo.std_dev**2+bar.std_dev**2)**0.5)

def flatMu(foo):
    return copyFlat(foo).mean()

def flatStd(foo):
    return copyFlat(foo).std()

def muStd(foo):
    return copyFlat(foo).std()/(len(foo)**0.5)

# Takes an array with uncertainties and
# clones it withouth the uncertainties
def copyFlat(bar):
    k = len(bar)
    flat = np.zeros(k)
    for j in range(0,k):
        flat[j] = bar[j].nominal_value if isinstance(bar[j], uncertainties.UFloat) else bar[j]
    return flat

def scaleSplitting(seg, start, end):
    singleElement = (end-start)/seg
    splitArray = np.array([start])
    for j in range (1, seg+1):
        splitArray = np.append(splitArray, j*singleElement)
    return splitArray
```
